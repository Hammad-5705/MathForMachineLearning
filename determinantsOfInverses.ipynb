## 1. Given the invertible matrix $A = \begin{pmatrix} 5 & 3 \\ 3 & 2 \end{pmatrix}$, find $\det(A^{-1})$ without explicitly computing the inverse matrix $A^{-1}$.

```python
import numpy as np
```

```python
A = np.array([
    [5,3],
    [3,2]
])

det_A = np.linalg.det(A)
if det_A != 0:
    det_A_inv = 1/det_A

    print(np.round(det_A_inv))
```

## 2. Let $B$ be a $3 \times 3$ invertible matrix with $\det(B) = -8$. What is the value of $\det((2B)^{-1})$?

```python
det_B = -8


# # Apply the scalar multiplication property.
# For an n x n matrix B and scaler c det(cB) = c^n(det(B))
det_2B = np.power(2,3) * det_B 

det_2B_inv = 1/det_2B

print(det_2B_inv)
```

## 3. If $A$ and $B$ are invertible $4 \times 4$ matrices with $\det(A) = 3$ and $\det(B) = -2$, calculate $\det(A^{-1} B A)$.

## 4. Suppose $M$ is an $n \times n$ orthogonal matrix, meaning $M^{-1} = M^T$. Using properties of determinants and inverses, what are the only two possible values for $\det(M)$?

**Definition of an orthogonal matrix:**
   Since $M$ is orthogonal, its inverse is equal to its transpose:
   $$M^{-1} = M^T$$

**Use the inverse property:**
   By definition of an inverse, $M \cdot M^{-1} = I$ (where $I$ is the $n \times n$ identity matrix). Substituting $M^T$ for $M^{-1}$ gives:
   $$M \cdot M^T = I$$

**Take the determinant of both sides:**
   $$\det(M \cdot M^T) = \det(I)$$

**Apply determinant properties:**
   * **Product rule:** $\det(M \cdot M^T) = \det(M) \cdot \det(M^T)$
   * **Transpose rule:** $\det(M^T) = \det(M)$
   * **Identity matrix:** $\det(I) = 1$

   Substituting these into the equation yields:
   $$\det(M) \cdot \det(M) = 1$$
   $$(\det(M))^2 = 1$$

**Solve for $\det(M)$:**
   Taking the square root of both sides gives:
   $$\det(M) = \pm 1$$


## 5. For what value(s) of $x$ does the matrix $C = \begin{pmatrix} x & 4 \\ 2 & x-2 \end{pmatrix}$ fail to have an inverse, and when $x = 5$, what is $\det(C^{-1})$?

```python
import sympy as sp

x= sp.symbols('x')

C= sp.Matrix([
    [x,4],
    [2,x-2]
])
```

```python
no_inv = sp.solve(C.det(),x)
print(f"Matrix C has no inverse at {no_inv}")
```

```python
# Substitute x=5
C_at_5 = C.subs(x,5)
C_at_5_inv = C_at_5.inv()
det_C_at_5_inv = C_at_5_inv.det()

print(det_C_at_5_inv)
```
