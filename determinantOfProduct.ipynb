## Let $A = \begin{pmatrix} 2 & 1 \\ 5 & 3 \end{pmatrix}$ and $B = \begin{pmatrix} 4 & -1 \\ 2 & 1 \end{pmatrix}$. Without computing the matrix product $AB$, find $\det(AB)$

the determinant product rule, $\det(AB) = \det(A)\det(B)$, which holds for all square matrices of the same size


```python
import numpy as np
```

```python
A = np.array([
    [2,1],
    [5,3]
])

B = np.array([
    [4,-1],
    [2,1]
])
```

```python
det_A = np.linalg.det(A)
det_B = np.linalg.det(B)

det_AB = det_A*det_B
print(np.round(det_AB))
```

## 2. If $A$ and $B$ are $3 \times 3$ matrices with $\det(A) = -4$ and $\det(B) = 5$, calculate $\det(A^2 B^3)$.


   For square matrices of the same size, the determinant of a product equals the product of their determinants:
   $$\det(A^2 B^3) = \det(A^2) \cdot \det(B^3)$$


   Raising a matrix to a power raises its determinant to the same power:
   $$\det(A^2) \cdot \det(B^3) = [\det(A)]^2 \cdot [\det(B)]^3$$


```python
det_A2 = np.square(-4)
det_B3 = np.power(5,3)

det_A2B3 = det_A2 * det_B3
print(np.round(det_A2B3))
```

## 3. Let $A$ be a $4 \times 4$ matrix such that $\det(A) = 3$. Find the value of $\det(2A \cdot A^T)$.

To find the determinant of $2A \cdot A^T$, use the standard properties of determinants for an $n \times n$ matrix (where $n = 4$):

1. **Scalar Multiplication Property:** Factor out the scalar $2$ by raising it to the power of the matrix's dimension ($n = 4$):
   $$\det(2A \cdot A^T) = 2^4 \cdot \det(A \cdot A^T)$$

2. **Product Property:** Split the determinant of the matrix product into the product of their individual determinants:
   $$2^4 \cdot \det(A \cdot A^T) = 16 \cdot \det(A) \cdot \det(A^T)$$

3. **Transpose Property:** The determinant of a transposed matrix is equal to the determinant of the original matrix ($\det(A^T) = \det(A)$):
   $$16 \cdot \det(A) \cdot \det(A) = 16 \cdot (3) \cdot (3) = 144$$


## 4. Suppose $A$ and $B$ are $n \times n$ matrices where $AB = I_n$ (the identity matrix). If $\det(A) = \frac{1}{7}$, what must $\det(B)$ be?

Take the determinant of both sides of $AB = I_n$:
   $$\det(AB) = \det(I_n)$$

Apply the product and identity properties:
   $$\det(A)\det(B) = 1$$

Substitute $\det(A) = \frac{1}{7}$:
   $$\frac{1}{7} \cdot \det(B) = 1$$

Multiply both sides by $7$:
   $$\det(B) = 7$$


## 5. Two $3 \times 3$ matrices $P$ and $Q$ satisfy $\det(PQ) = 0$. If $P$ is known to be non-singular with $\det(P) = 6$, what does this imply about the singularity and determinant of $Q$?


**Apply the Determinant Product Rule**  
   For any two square matrices $P$ and $Q$ of the same order ($3 \times 3$), the determinant of their product equals the product of their individual determinants:
   $$\det(PQ) = \det(P) \cdot \det(Q)$$

**Substitute the Given Values**  
   Given $\det(PQ) = 0$ and $\det(P) = 6$:
   $$0 = 6 \cdot \det(Q)$$

**Solve for $\det(Q)$**  
   Divide both sides by $6$:
   $$\det(Q) = \frac{0}{6} = 0$$

**Determine Singularity**  
   A square matrix is **singular** (non-invertible) if and only if its determinant is $0$. Since $\det(Q) = 0$, **$Q$ must be singular**.
