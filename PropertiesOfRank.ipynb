### Suppose a linear transformation $L: \mathbb{R}^4 \to \mathbb{R}^4$ maps a non-zero vector $\mathbf{v}$ to the zero vector $\mathbf{0}$. What can you conclude about the singularity of $L$ and the maximum possible value of its rank?

Because \(L\) maps a non-zero vector to \(\mathbf{0}\), its kernel has dimension at least \(1\), which makes \(L\) singular (non-invertible) and limits its maximum rank to \(3\) by the Rank-Nullity Theorem (\(4 - 1 = 3\)).
