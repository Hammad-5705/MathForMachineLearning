## 1. Suppose a linear transformation $L: \mathbb{R}^4 \to \mathbb{R}^4$ maps a non-zero vector $\mathbf{v}$ to the zero vector $\mathbf{0}$. What can you conclude about the singularity of $L$ and the maximum possible value of its rank?

Because $L$ maps a non-zero vector to $\mathbf{0}$, its kernel has dimension at least $1$, which makes $L$ singular (non-invertible) and limits its maximum rank to $3$ by the Rank-Nullity Theorem ($4 - 1 = 3$).

**Rank-Nullity Theorem:**     $\operatorname{rank}(L) + \operatorname{nullity}(L) = \dim(V)$


## 2. For what value(s) of $k$ does the linear transformation $T(x, y) = (2x + ky, 4x + 6y)$ become singular?

### A linear transformation is singular (non-invertible) if and only if the determinant of its matrix representation is \(0\).

```python
# define variable K
import sympy as sp
k = sp.symbols('K')
```

```python
A = sp.Matrix([
    [2,k],
    [4,6]
])

det_A = A.det()
res = sp.solve(det_A, k)

print(f"Singular when k = {res[0]}")
```
