## 1. Suppose a linear transformation $L: \mathbb{R}^4 \to \mathbb{R}^4$ maps a non-zero vector $\mathbf{v}$ to the zero vector $\mathbf{0}$. What can you conclude about the singularity of $L$ and the maximum possible value of its rank?

Because $L$ maps a non-zero vector to $\mathbf{0}$, its kernel has dimension at least $1$, which makes $L$ singular (non-invertible) and limits its maximum rank to $3$ by the Rank-Nullity Theorem ($4 - 1 = 3$).

**Rank-Nullity Theorem:**     $\operatorname{rank}(L) + \operatorname{nullity}(L) = \dim(V)$


## 2. For what value(s) of $k$ does the linear transformation $T(x, y) = (2x + ky, 4x + 6y)$ become singular?

### A linear transformation is singular (non-invertible) if and only if the determinant of its matrix representation is \(0\).

```python
# define variable K
import sympy as sp
k = sp.symbols('K')
```

```python
A = sp.Matrix([
    [2,k],
    [4,6]
])

det_A = A.det()
res = sp.solve(det_A, k)

print(f"Singular when k = {res[0]}")
```

## 4. Consider a linear transformation $T: \mathbb{R}^3 \to \mathbb{R}^3$ that projects every vector in 3D space onto the $xy$-plane. Find the rank of $T$ and determine if it is singular.

$$T\left(\begin{bmatrix} x \\ y \\ z \end{bmatrix}\right) = \begin{bmatrix} x \\ y \\ 0 \end{bmatrix}$$

To find the standard matrix representation $A$, apply $T$ to the standard basis vectors of $\mathbb{R}^3$:

$$T(\mathbf{e}_1) = T\left(\begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}\right) = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}, \quad T(\mathbf{e}_2) = T\left(\begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}\right) = \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}, \quad T(\mathbf{e}_3) = T\left(\begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}\right) = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$$

Combining these images as columns gives the standard matrix $A$:

$$A = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 0 \end{bmatrix}$$




The **rank** of a linear transformation is the dimension of its image (range), which equals the number of linearly independent columns (or pivot positions) in its matrix $A$:

* The first two columns of $A$, $\begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}$ and $\begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix}$, are linearly independent and span the entire $xy$-plane (a $2$-dimensional subspace of $\mathbb{R}^3$).
* The third column is the zero vector $\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$, which contributes no dimension.

$$\text{rank}(T) = 2$$



### Determining Singularity
A linear operator $T: \mathbb{R}^n \to \mathbb{R}^n$ is **singular** (not invertible) if it does not have full rank ($\text{rank}(T) < n$), or equivalently, if its kernel (null space) contains non-zero vectors.

* **By the Rank-Nullity Theorem:**
  $$\text{nullity}(T) = n - \text{rank}(T) = 3 - 2 = 1 > 0$$
* **Kernel Analysis:** Every vector along the $z$-axis of the form $\begin{bmatrix} 0 \\ 0 \\ z \end{bmatrix}$ maps to the zero vector $\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$. Since $T$ is not injective (one-to-one), the original $z$-coordinate cannot be recovered from the projected output.
* **By Determinant:**
  $$\det(A) = 1 \cdot 1 \cdot 0 = 0$$

Therefore, $T$ is **singular**.
