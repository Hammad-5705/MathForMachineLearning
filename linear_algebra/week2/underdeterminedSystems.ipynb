## 1. Find the general solution to the following underdetermined system by expressing $x$ and $y$ in terms of the free variable $z$:
   $$x + 2y + 3z = 7$$
   $$2x + 5y + 8z = 16$$

```python
import sympy as sp

x,y,z = sp.symbols('x y z')

eq1 = sp.Eq(x + 2*y + 3*z, 7)
eq2 = sp.Eq(2*x + 5*y + 8*z, 16)

sol = sp.linsolve([eq1,eq2], (x,y,z))
sp.pprint(sol)
```

## 2. Consider a consistent system of $2$ linear equations in $4$ unknowns ($x_1, x_2, x_3, x_4$). What is the minimum number of free variables this system must have?


The system must have a minimum of **$2$ free variables**.

### Explanation
For a consistent system of linear equations, the number of free variables is determined by the relationship:

$$\text{Free Variables} = (\text{Number of Unknowns}) - (\text{Number of Pivot Positions})$$

* **Number of unknowns:** $4$ ($x_1, x_2, x_3, x_4$).
* **Maximum pivot positions:** $2$. Because there are only $2$ equations, the coefficient matrix has $2$ rows. Each row can contain at most one pivot position (leading entry), which occurs when the two equations are linearly independent.

Subtracting the maximum number of pivots from the total number of unknowns gives the minimum number of free variables:

$$4 - 2 = 2$$

*(Note: If the two equations are dependent/multiples of each other, there is only $1$ pivot position, resulting in $4 - 1 = 3$ free variables.)*


## 3. Geometrically, what is the intersection of two non-parallel planes in $\mathbb{R}^3$, and how does this relate to a system of $2$ equations with $3$ variables?

```python
x, y, z = sp.symbols('x y z')

eq1 = sp.Eq(x + y + z, 2)
eq2 = sp.Eq(x - y + z, 0)

solution = sp.linsolve([eq1, eq2], (x, y, z))
sp.pprint(solution)
```

### Intersection of Two Non-Parallel Planes in $\mathbb{R}^3$

* **Geometric Meaning:** Two non-parallel planes in $\mathbb{R}^3$ always intersect along a **single infinite straight line**.
* **Algebraic Connection ($2$ Equations, $3$ Variables):**
  * Each plane is represented by one linear equation in $x$, $y$, and $z$.
  * Because the planes are not parallel, their normal vectors are linearly independent, resulting in **2 pivot variables** and **1 free variable** (e.g., $z = t$).
  * Having **1 free variable** gives the solution set **1 degree of freedom**, which produces **infinitely many solutions** forming the parametric equation of a line:
    $$\mathbf{r}(t) = \mathbf{r}_0 + t\mathbf{v}$$
* **Example (`{(1 - z, 1, z)}`):** Letting $z = t$ be the free variable yields $(x, y, z) = (1 - t, 1, t)$, a line passing through $(1, 1, 0)$ in the direction $\langle -1, 0, 1 \rangle$.


## 4. Write the general solution of the system below as a linear combination of vectors using parameters $s$ and $t$:
   $$x_1 - 2x_2 + x_3 + 3x_4 = 4$$
   $$x_3 - 2x_4 = 1$$

```python
x1, x2, x3, x4, s, t = sp.symbols('x1 x2 x3 x4 s t')


eq1 = sp.Eq(x1 - 2 * x2 + x3 + 3 * x4, 4)
eq2 = sp.Eq(x3 - 2 * x4, 1)


sol = sp.solve((eq1, eq2), (x1, x3))

sp.pprint(sol)
```

```python
# substitute x2 = s and x4 = t
general_vec = sp.Matrix([
    sol[x1].subs({x2: s, x4: t}),
    s,
    sol[x3].subs({x2: s, x4: t}),
    t,
])

print(general_vec)
```

```python
# decompose into vectors
v_const = general_vec.subs({s: 0, t: 0})
v_s = sp.diff(general_vec, s)
v_t = sp.diff(general_vec, t)

# Build an unevaluated linear combination expression
linear_comb = sp.MatAdd(
    v_const,
    sp.MatMul(s, v_s, evaluate=False),
    sp.MatMul(t, v_t, evaluate=False),
    evaluate=False,
)

sp.pprint(linear_comb)
```

## 5. Can a system of linear equations with more variables than equations ever be inconsistent (have no solution)? Provide a simple $2 \times 3$ example to justify your answer.

**Yes, a system of linear equations with more variables than equations can be inconsistent** when the equations represent parallel planes that contradict each other.

### 2 × 3 Example

$$
\begin{aligned}
x + y + z &= 1 \\
2x + 2y + 2z &= 5
\end{aligned}
$$

* **Justification:** Multiplying the first equation by $2$ gives $2x + 2y + 2z = 2$, which contradicts the second equation ($2x + 2y + 2z = 5$). Since $2 \neq 5$, no solution exists.


```python

x, y, z = sp.symbols('x y z')


eq1 = sp.Eq(x + y + z, 1)
eq2 = sp.Eq(2*x + 2*y + 2*z, 5)


sol = sp.linsolve([eq1, eq2], (x, y, z))

sp.pprint(sol)
```
