## 1. State the exact conditions that distinguish a matrix in *Reduced Row Echelon Form (RREF)* from a matrix that is only in *Row Echelon Form (REF)*.

To be in **Reduced Row Echelon Form (RREF)** rather than just **Row Echelon Form (REF)**, a matrix must meet all REF conditions plus **two exact distinguishing conditions**:

1. **Leading 1s:** The first nonzero entry (pivot) in every nonzero row must equal **1** *(in REF, pivots can be any nonzero number)*.
2. **Zeros above pivots:** Every leading **1** must be the **only nonzero entry in its column**, meaning all entries *above* each pivot are also **0** *(in REF, only entries below pivots must be 0)*.

## 3. Reduce the following matrix to *Row Echelon Form (REF)*:
   $$M = \begin{pmatrix} 1 & 2 & -1 & 3 \\ 2 & 5 & 1 & 8 \\ 3 & 7 & 0 & 11 \end{pmatrix}$$

```python
import sympy as sp

M = sp.Matrix([
    [1,2,-1,3],
    [2,5,1,8],
    [3,7,0,11]
])
```

```python
res = M.echelon_form()
sp.pprint(res)
```

## 4. Continue reducing the matrix $M$ from Question 3 until it is in *Reduced Row Echelon Form (RREF)*.

```python
res, piv = M.rref()

sp.pprint(res)
sp.pprint(piv)
```

## 5. Given the augmented matrix in RREF below, write down the complete solution set of the corresponding linear system:
   $$\left(\begin{array}{cccc|c} 1 & 0 & 3 & 0 & 5 \\ 0 & 1 & -2 & 0 & -1 \\ 0 & 0 & 0 & 1 & 4 \end{array}\right)$$

```python
M = sp.Matrix([
    [1,0,3,0,5],
    [0,1,-2,0,-1],
    [0,0,0,1,4]
])
w,x,y,z = sp.symbols('w,x,y,z')
```

```python
res = sp.linsolve(M, w,x,y,z)
sp.pprint(res)
```

```python
sol_vec = sp.Matrix(next(iter(res)))
sp.pprint(sol_vec)
```

```python
cons_vec = sol_vec.subs(y, 0)
sp.pprint(cons_vec)
```

```python
y_vec = sol_vec.diff(y)
sp.pprint(y_vec)
```

```python
ans = sp.MatAdd(
    cons_vec,
    sp.MatMul(y, y_vec, evaluate=False),
    evaluate= False
)
sp.pprint(ans)
```

```python
var_vec = sp.Matrix([w,x,y,z])
sp.pprint(var_vec)
```

```python
eq = sp.Eq(var_vec, ans)
sp.pprint(eq)
```
