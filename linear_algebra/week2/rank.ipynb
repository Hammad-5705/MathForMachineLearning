## 2. Let $B$ be a $5 \times 3$ matrix. What is the maximum possible rank of $B$, and what does it mean for its columns if $B$ achieves this maximum rank?

The maximum possible rank of a $5 \times 3$ matrix $B$ is **$3$**, which means that **its $3$ columns are linearly independent**.

### Why the maximum rank is 3
The rank of a matrix cannot exceed the smaller of its row count or column count:
$$\text{rank}(B) \le \min(m, n)$$

For a $5 \times 3$ matrix ($m = 5$ rows and $n = 3$ columns):
$$\text{rank}(B) \le \min(5, 3) = 3$$

### What achieving maximum rank means for the columns
The rank of a matrix equals the number of linearly independent columns (the dimension of its column space). When $\text{rank}(B) = 3$:
* **Full column rank:** All $3$ columns are linearly independent, meaning no column can be written as a linear combination of the other two.
* **Trivial null space:** The only solution to the homogeneous equation $B\mathbf{x} = \mathbf{0}$ is the zero vector $\mathbf{x} = \mathbf{0}$.
* **Span in $\mathbb{R}^5$:** The $3$ columns form a basis for a $3$-dimensional subspace inside $\mathbb{R}^5$ (they do not span all of $\mathbb{R}^5$, since $5$ vectors are needed for that).


## 3. If $A$ is a $4 \times 4$ matrix with $\text{rank}(A) = 3$, what is the value of $\det(A)$, and what is the dimension of the null space (kernel) of $A$?

For a $4 \times 4$ matrix $A$ with $\text{rank}(A) = 3$, the determinant is **$\det(A) = 0$** and the dimension of the null space is **$\dim(\text{Nul}(A)) = 1$**.

### Value of the determinant
* An $n \times n$ matrix is invertible if and only if it has full rank ($\text{rank}(A) = n$).
* Since $A$ is a $4 \times 4$ matrix with $\text{rank}(A) = 3 < 4$, its rows and columns are linearly dependent, making $A$ singular (non-invertible).
* The determinant of any singular square matrix is **$0$**.

### Dimension of the null space (nullity)
* By the **Rank-Nullity Theorem**, for any $m \times n$ matrix $A$, the rank and nullity sum to the number of columns ($n$):
  $$\text{rank}(A) + \text{nullity}(A) = n$$
* Substituting $n = 4$ and $\text{rank}(A) = 3$:
  $$3 + \text{nullity}(A) = 4$$
  $$\text{nullity}(A) = 4 - 3 = 1$$


## 4. Prove or explain why $\text{rank}(A) = \text{rank}(A^T)$ for any $m \times n$ matrix $A$.

For any $m \times n$ matrix $A$, **$\text{rank}(A) = \text{rank}(A^T)$** because the **row rank** of a matrix always equals its **column rank**, and transposing $A$ swaps its rows and columns.

---

### 1. Subspace & Dimension Explanation
* **Column space and row space:** By definition, $\text{rank}(A)$ is the dimension of the column space of $A$, denoted $\dim(\text{col}(A))$. The row space of $A$ is the column space of $A^T$, so $\dim(\text{row}(A)) = \dim(\text{col}(A^T)) = \text{rank}(A^T)$.
* **Row rank equals column rank:** In linear algebra, the maximum number of linearly independent rows (row rank) always equals the maximum number of linearly independent columns (column rank).
* **Effect of transpose:** Swapping rows and columns turns the row space of $A$ into the column space of $A^T$, preserving that dimension: $\text{rank}(A) = \text{rank}(A^T)$.

---

### 2. Proof via Invertible Matrices (Normal Form)
Any $m \times n$ matrix $A$ of rank $r$ can be reduced via elementary row and column operations to the block normal form:

$$P A Q = D = \begin{bmatrix} I_r & 0 \\ 0 & 0 \end{bmatrix}$$

where $P$ is an invertible $m \times m$ matrix, $Q$ is an invertible $n \times n$ matrix, and $I_r$ is the $r \times r$ identity matrix.

1. Multiplying by invertible matrices preserves rank, so:
   $$\text{rank}(A) = \text{rank}(PAQ) = \text{rank}(D) = r$$
2. Take the transpose of both sides:
   $$(PAQ)^T = D^T \implies Q^T A^T P^T = D^T$$
3. Since $P$ and $Q$ are invertible, $P^T$ and $Q^T$ are also invertible.
4. Because $D$ is block-diagonal with $I_r$ in the upper-left corner, $D^T$ has the exact same structure and the same number of pivot ones ($r$), so $\text{rank}(D^T) = r$.
5. Since multiplying by invertible matrices preserves rank:
   $$\text{rank}(A^T) = \text{rank}(Q^T A^T P^T) = \text{rank}(D^T) = r = \text{rank}(A)$$

---

### 3. Proof via Gram Matrices ($A^T A$ and $A A^T$) over $\mathbb{R}$
For a real matrix $A$, the null space of $A^T A$ equals the null space of $A$:
* If $Ax = 0$, then $A^T A x = A^T(0) = 0$.
* Conversely, if $A^T A x = 0$, then $x^T A^T A x = \|Ax\|_2^2 = 0$, which forces $Ax = 0$.

By the Rank-Nullity Theorem applied to $A$ ($\text{rank}(A) + \text{nullity}(A) = n$) and to the $n \times n$ matrix $A^T A$ ($\text{rank}(A^T A) + \text{nullity}(A^T A) = n$), equality of nullities gives:

$$\text{rank}(A) = \text{rank}(A^T A)$$

Replacing $A$ with $A^T$ yields $\text{rank}(A^T) = \text{rank}(A A^T)$. Because $A^T A$ and $A A^T$ share the same nonzero eigenvalues (and therefore the same rank), it follows that:

$$\text{rank}(A) = \text{rank}(A^T)$$


## 5. For what value(s) of $a$ does the matrix $C = \begin{pmatrix} 1 & 2 & a \\ 2 & 4 & 6 \\ 3 & 6 & 9 \end{pmatrix}$ have a rank of $1$?

A non-zero matrix has a rank of \(1\) if and only if all of its rows are scalar multiples of a single non-zero row.

The matrix \(C\) has a rank of \(1\) when \(a = 3\).
