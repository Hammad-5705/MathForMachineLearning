## 1. Solve the following system of linear equations and state why the coefficient matrix is non-singular:
   $$2x + 3y = 8$$
   $$5x - y = 3$$

```python
import sympy as sp

x = sp.symbols('x')
y = sp.symbols('y')
```

```python
eq_1 = sp.Eq(2*x + 3*y, 8)
eq_2 = sp.Eq(5*x - y, 3)

sol = sp.solve([eq_1, eq_2], [x,y])

print(sol)
```

### A square matrix is non-singular (invertible) if and only if its determinant is not equal to zero.

```python
A = sp.Matrix([
    [2,3],
    [5,-1]
])

print(f"As Det of A is {A.det()} so the system is non-singular and has a unique solution.")
```

## 2. Given the matrix equation $A\mathbf{x} = \mathbf{b}$ where $A = \begin{pmatrix} 1 & 2 & 1 \\ 0 & 3 & -1 \\ 2 & 1 & 4 \end{pmatrix}$ and $\mathbf{b} = \begin{pmatrix} 4 \\ 5 \\ 7 \end{pmatrix}$, verify that $A$ is non-singular and find the unique solution vector $\mathbf{x}$.

```python
import sympy as sp

A = sp.Matrix([
    [1, 2, 1],
    [0, 3, -1],
    [2, 1, 4]
])
B = sp.Matrix([4, 5, 7])
```

As $AX = b$, so that $X = A^{-1}b$


```python
# Verify A is non-singular
det_A = A.det()
print("The Determinant of A is", det_A, " So its Non-Singular and has a unique solution.") 
```

```python
X = A.inv() @ B
print("Solution vector x:")
sp.pprint(X)
```

## 3. For what value(s) of $k$ does the system $\begin{pmatrix} k & 3 \\ 4 & k+1 \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 5 \\ 2 \end{pmatrix}$ have a unique solution?

```python
k = sp.symbols('k')

M = sp.Matrix([
    [k, 3],
    [4, k+1]
])
```

```python
det_M = M.det()
print("The Determinant of M is", det_M, " which is a function of k. The system is non-singular for values of k that make the determinant non-zero.")
```

```python
sol = sp.solve(det_M, k)
print(f"The system has unique solution for all values of K except : {sol}")
```

## 4. Geometrically, what does the unique solution of a $3 \times 3$ non-singular system of linear equations represent in $\mathbb{R}^3$?

Geometrically, the unique solution of a $3 \times 3$ non-singular system of linear equations represents **the single point of intersection of three planes in $\mathbb{R}^3$**.

### Key Geometric Breakdown

* **Each equation defines a plane:** A linear equation in three variables ($ax + by + cz = d$) represents a flat, two-dimensional plane in three-dimensional space ($\mathbb{R}^3$).
* **Non-singular orientation:** A non-singular coefficient matrix ($\det(A) \neq 0$) means the normal vectors of the three planes are linearly independent. None of the planes are parallel, nor do they all intersect along a single common line.
* **Intersection at a unique point:** Any two of the planes intersect along a straight line, and the third plane slices across that line at exactly **one unique point $(x, y, z)$**, which satisfies all three equations simultaneously.
