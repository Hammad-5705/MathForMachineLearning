## 1. Solve the following system of equations step-by-step using the *Gaussian Elimination algorithm* (forward elimination to upper triangular form, followed by back substitution):
   $$x + y + 2z = 9$$
   $$2x + 4y - 3z = 1$$
   $$3x + 6y - 5z = 0$$

```python
import sympy as sp

M = sp.Matrix([
    [1,1,2,9],
    [2,4,-3,1],
    [3,6,-5,0]
])
x,y,z = sp.symbols('x y z')
```

```python
M = M.elementary_row_op('n->n+km', row1=1, k=-2, row2=0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1=2, k=-3, row2=0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=1, k=sp.Rational(1,2))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1= 2, k= -3, row2=1)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row= 2, k= sp.Rational(-2,1))
sp.pprint(M)
```

```python
eq3 = sp.Eq(M[2,2]*z, M[2,3])
sp.pprint(eq3)
```

```python
eq2 = sp.Eq(M[1,1]*y + M[1,2]*z, M[1,3])
sp.pprint(eq2)
```

```python
eq1 = sp.Eq(M[0,0]*x + M[0,1]*y + M[0,2]*z, M[0,3])
sp.pprint(eq3)
```

```python
sol_z = sp.solve(eq3, z)[0]
sol_y = sp.solve(eq2.subs(z,sol_z),y)[0]
sol_x = sp.solve(eq1.subs({z: sol_z, y: sol_y}), x)[0] 

print(f"x = {sol_x}\ny = {sol_y}\nz = {sol_z}")
```

## 2. During the forward elimination step on the system below, a zero appears in the pivot position. Show how *partial pivoting* (row swapping) resolves this issue and complete the reduction to REF:
   $$2y + 3z = 7$$
   $$x + y + z = 4$$
   $$3x - y + 2z = 5$$

```python
M = sp.Matrix([
    [0,2,3,7],
    [1,1,1,4],
    [3,-1,2,5]
])

x,y,z = sp.symbols('x y z')

sp.pprint(M)
```

```python
M.row_swap(0,2)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=0, k = sp.Rational(1,3))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1=1, k= -1, row2=0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=1, k= sp.Rational(3,4))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row= 2, k= sp.Rational(1,2))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1= 2, k=-1, row2= 1)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=2, k= sp.Rational(1,1.25))
sp.pprint(M)
```

```python
eq3 = sp.Eq(M[2,2]*z, M[2,3])
eq2 = sp.Eq(M[1,1]*y + M[1,2]*z , M[1,3])
eq1 = sp.Eq(M[0,0]*x + M[0,1]*y +M[0,2]*z, M[0,3])
```

```python
sp.pprint(eq1)
sp.pprint(eq2)
sp.pprint(eq3)
```

```python
sol_z = sp.solve(eq3, z)[0]
sol_y = sp.solve(eq2.subs(z, sol_z), y)[0]
sol_x = sp.solve(eq1.subs({y:sol_y, z:sol_z}), x)[0]
```

```python
print(f"x = {sol_x}, y = {sol_y}, z = {sol_z}")
```

## Explain why partial pivoting (choosing the row with the largest absolute value in the pivot column) is used in computer implementations of Gaussian elimination even when the current pivot is non-zero (e.g., $10^{-16}$).

**Partial pivoting** (swapping rows to place the largest absolute value in the pivot column on top) is essential in computer implementations of Gaussian elimination—even when the current pivot is non-zero (e.g., $10^{-16}$)—to maintain **numerical stability** and prevent **floating-point rounding errors**.

#### Core Reasons:
1. **Prevents Massive Multipliers:** 
   Gaussian elimination computes row multipliers as $m = \frac{a_{ik}}{a_{kk}}$. Dividing by a tiny pivot like $10^{-16}$ produces a huge multiplier ($\sim 10^{16}$). Placing the largest value on top guarantees $|m| \le 1$.
2. **Avoids Swamping (Loss of Significance):** 
   Because computers use finite floating-point precision (~15–17 significant digits in 64-bit IEEE 754), multiplying a row by $\sim 10^{16}$ and subtracting it from another row causes the large numbers to completely overwhelm and erase the smaller original digits.
3. **Prevents Catastrophic Cancellation:** 
   Subtracting nearly equal large numbers during back-substitution wipes out remaining significant digits, resulting in severe inaccuracies or completely wrong solutions.
4. **Low Computational Cost:** 
   Scanning a single column for the maximum absolute value adds negligible overhead while dramatically improving accuracy.


## 4. Apply Gaussian elimination to the augmented matrix below to determine for which value(s) of $k$ the system has no solution:
   $$\left(\begin{array}{ccc|c} 1 & 1 & 1 & 2 \\ 2 & 3 & 2 & 5 \\ 3 & 4 & k & 8 \end{array}\right)$$

```python
M = M.elementary_row_op('n->kn', row=1, k= sp.Rational(1,2))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1= 1, k= -1, row2= 0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row= 1, k= 2)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1= 2, k= sp.Rational(1,-3), row2= 1)
sp.pprint(M)
```

## 5. How many total pivot positions exist in an $m \times n$ matrix whose rank is $r$, and how do these pivots identify the basic (leading) variables versus the free variables in a linear system?

An $m \times n$ matrix of rank $r$ has **$r$ total pivot positions**, which correspond directly to the **$r$ basic (leading) variables**, leaving **$n - r$ free variables**.

### 1. Total Number of Pivot Positions
* **Rank ($r$):** By definition, the rank of a matrix is the number of nonzero rows in its row echelon form (or reduced row echelon form).
* **Pivot Count:** Since every nonzero row in row echelon form contains exactly one leading entry (a pivot), the total number of pivot positions in the matrix is always **$r$**.
* **Bounds on $r$:** Because a matrix cannot have more pivots than rows or columns, $r$ always satisfies $r \le \min(m, n)$.

### 2. Identifying Basic vs. Free Variables
In an augmented matrix representing a consistent linear system with $n$ variables (corresponding to the $n$ columns of the coefficient matrix):

| Variable Type | Column Criterion | Total Count | Behavior in Solution |
| :--- | :--- | :--- | :--- |
| **Basic (Leading) Variables** | Correspond to **pivot columns** (columns that contain a pivot position). | **$r$** | **Dependent:** Their values are uniquely determined once the free variables are chosen. |
| **Free Variables** | Correspond to **non-pivot columns** (columns that do *not* contain a pivot position). | **$n - r$** | **Independent:** Can be assigned any real number as arbitrary parameters. |

#### Example Setup
Consider a consistent system with $n = 4$ variables ($x_1, x_2, x_3, x_4$) and rank $r = 2$, whose coefficient matrix reduces to:

$$\begin{bmatrix} 1 & 2 & 0 & -3 \\ 0 & 0 & 1 & 5 \\ 0 & 0 & 0 & 0 \end{bmatrix}$$

* **Pivots:** Located in row 1, column 1 and row 2, column 3 ($r = 2$ pivots total).
* **Basic variables:** **$x_1$ and $x_3$** (since columns 1 and 3 contain pivots).
* **Free variables:** **$x_2$ and $x_4$** (since columns 2 and 4 lack pivots; $n - r = 4 - 2 = 2$).
