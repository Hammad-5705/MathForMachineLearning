### 1. Let $T: \mathbb{R}^3 \to \mathbb{R}^3$ be a linear transformation represented by the matrix $A = \begin{pmatrix} 1 & 2 & 3 \\ 2 & 4 & 6 \\ 1 & 1 & 1 \end{pmatrix}$. Determine the rank of $T$ and state whether $T$ is singular or non-singular.

```python
import numpy as np
```

## Rank:
Row Echelon Form 

This is the most common and reliable method for any matrix of size m X n

```python
# matrix
A = [
    [1,2,3],
    [2,4,6],
    [1,1,1]
]
```

```python
# function to convert matrix into echelon form
def echelon_form(m):
    # first row pivot and zero under it
    if m[0][0] != 1:
        pivot = m[0][0]
        for i in range(3):
            m[0][i] = m[0][i] / pivot

        for i in range(1,3):
            for j in range(3):
                m[i][j] = (m[i][j]*m[i][j]) - m[i][j]
        

    # Second Row pivot and zero under it
    if m[2][2] != 1:
        pivot = m[2][2]
        for i in range(3):
            m[2][i] = m[2][i] / pivot

        for i in range(2,3):
            for j in range(3):
                m[i][j] = (m[i][j]*m[i][j]) - m[i][j]

    # Third Row pivot and zero under it
    if m[2][2] !=1:
        pivot = m[2][2]
        for i in range(3):
            m[2][i] = m[2][1] / pivot

    return m
```

```python
# function to get rank of matrix
def get_rank(m):
    if np:
        return np.linalg.matrix_rank(m)
    rank = 0
    for i in range(len(m)):
        if m[i][i] == 1:
            rank +=1

    return rank
```

```python
echelon = echelon_form(A)
print(echelon)
```

```python
rank = get_rank(echelon)
print(rank)
```

```python
print("As the rank of the matrix is less than the number of rows, the matrix is singular.")
```

## Determinant Test

* **Singular:** If $\det(A) = 0$, the matrix is singular (it has no inverse and squishes space into a lower dimension).
* **Non-singular:** If $\det(A) \neq 0$, the matrix is non-singular (it is invertible and preserves dimension).


```python
def get_determinant(m):
    if np:
        return np.linalg.det(m)
    return ((m[0][0]*m[1][1]*m[2][2]) + (m[1][0]*m[2][1]*m[0][2]) + (m[2][0]*m[0][1]*m[1][2])) - ((m[0][2]*m[1][1]*m[2][0]) + (m[0][1]*m[1][0]*m[2][2]) + (m[0][0]*m[2][1]*m[1][2]))
```

```python
det = get_determinant(A)
print(det)
```

```python
print("As determinant is 0, the matrix is singular.")
```

## Rows & col dependencies Test

### A square matrix A of size n × n is singular if its rows (and equivalently its columns) are linearly dependent

• Zero row or column: An entire row or column is all zeros.

• Identical rows or columns: Two rows or two columns are exactly the same.

• Scalar multiples: One row or column is a multiple of another (for example, Row 2 equals 3 × Row 1).

• Linear combination: Any row or column can be formed by adding or scaling other rows or columns.


```python
def dependencies(m):
    if m[0][0]*2 == m[1][0] and m[0][1]*2 == m[1][1] and m[0][2]*2 == m[1][2]:
        return True
```

```python
is_dep = dependencies(A)
if is_dep:
    print("As the rows of the matrix are linearly dependent, the matrix is singular.")
else:
    print("As the rows of the matrix are linearly independent, the matrix is non-singular.")
```

## Eigenvalues Test

if All Eigenvalues are Non-Zero the matrix is Non-Singular 

if At least 1 Eigenvalues is zero the matrix is Singular

```python
def get_eigan_values(m):
   return np.linalg.eigvals(m)
```

```python
eigen_val = get_eigan_values(A)
is_singular = np.any(np.isclose(eigen_val, 0))
if is_singular:
    print("As one of the eigenvalues is 0, the matrix is singular.")
else:
    print("As none of the eigenvalues is 0, the matrix is non-singular.")
```
