## 1. Write the exact Python code using numpy and numpy.linalg to define and solve the system:
   $$4x + y - z = 7$$
   $$x - 3y + 2z = -2$$
   $$2x + 2y + 5z = 11$$

```python
import numpy as np
```

```python
M = np.array([
    [4, 1, -1],
    [1, -3, 2],
    [2, 2, 5]
])

S = np.array([7, -2, 11])
```

```python
solution = np.linalg.solve(M, S)

print(np.round(solution))
```

## 4. Suppose A is a $4 \times 4$ matrix and b is a $4 \times 1$ vector. Explain the difference between executing np.linalg.solve(A, b) versus np.linalg.inv(A) @ b in terms of numerical accuracy and computational speed.

Both `np.linalg.solve(A, b)` and `np.linalg.inv(A) @ b` find the solution vector $x$ to the linear system $Ax = b$, but **`np.linalg.solve(A, b)` is faster and more numerically accurate**.

* **Computational Speed:** `np.linalg.solve(A, b)` solves $Ax = b$ directly using LU decomposition ($\approx \frac{2}{3}n^3$ operations). In contrast, `np.linalg.inv(A) @ b` computes the entire inverse matrix $A^{-1}$ before multiplying by $b$ ($\approx 2n^3$ operations, roughly 3x slower). While the wall-clock difference is negligible for a small $4 \times 4$ matrix, `solve()` is strictly more efficient and avoids extra function overhead.
* **Numerical Accuracy:** `np.linalg.solve(A, b)` minimizes floating-point rounding errors by solving directly for $x$. Computing $A^{-1}$ explicitly introduces extra rounding steps and significantly amplifies precision errors if $A$ is ill-conditioned (nearly singular).


## 5. What will np.linalg.matrix_rank(A) return for the array A = np.array([[1, 2, 3], [2, 4, 6], [0, 1, 0]])?

```python
A = np.array([[1, 2, 3], [2, 4, 6], [0, 1, 0]])

print(np.linalg.matrix_rank(A))
```

`np.linalg.matrix_rank(A)` will return **`2`**.

**Why:** The rank of a matrix is the number of linearly independent rows. In matrix $A$, the second row (`[2, 4, 6]`) is simply a scalar multiple ($2\times$) of the first row (`[1, 2, 3]`), meaning it can be eliminated to `[0, 0, 0]` using elementary row operations ($R_2 \leftarrow R_2 - 2R_1$). The remaining two rows (`[1, 2, 3]` and `[0, 1, 0]`) are linearly independent, leaving **2 non-zero rows**.
