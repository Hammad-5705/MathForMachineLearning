## 1. Determine whether the following singular system has *no solution* or *infinitely many solutions*:
   $$3x - 6y = 9$$
   $$-2x + 4y = -6$$

```python
import sympy as sp


x, y = sp.symbols('x y')


eq1 = sp.Eq(3*x - 6*y, 9)
eq2 = sp.Eq(-2*x + 4*y, -6)


solution = sp.linsolve([eq1, eq2], (x, y))
print(solution)
```

This output means y can be any real number, and x = 2y + 3 confirming infinitely many solutions.

## 4. Solve the following singular homogeneous system by expressing the solution in vector parametric form:
   $$\begin{pmatrix} 1 & 3 & -2 \\ 2 & 6 & -4 \\ -1 & -3 & 2 \end{pmatrix} \begin{pmatrix} x \\ y \\ z \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \\ 0 \end{pmatrix}$$

```python
A = sp.Matrix([
    [1,  3, -2],
    [2,  6, -4],
    [-1, -3,  2]
])


red_ech , pivot = A.rref()

dir_vec = A.nullspace()
s,t = sp.symbols('s t')
parameters = [s, t]

sol = s * dir_vec[0] + t * dir_vec[1]
sp.pprint(sol)
```

```python
v_const = sol.subs({s:0, t:0})
v_s = sp.diff(sol, s)
v_t = sp.diff(sol, t)

vec = sp.MatAdd(
    v_const,
    sp.MatMul(s, v_s, evaluate=False),
    sp.MatMul(t, v_t, evaluate=False),
    evaluate= False
)

sp.pprint(vec)
```

## 5. Explain why a singular system $A\mathbf{x} = \mathbf{b}$ can never have exactly two distinct solutions.

A linear system $A\mathbf{x} = \mathbf{b}$ represents straight lines (or flat planes) because all variables are only to the first power. If $\mathbf{x}_{1}$ and $\mathbf{x}_{2}$ are two distinct solutions, then any point on the straight line connecting them—such as $\mathbf{x}_1 + c(\mathbf{x}_2 - \mathbf{x}_1)$ for any real number $c$—is also a solution:

$$A(\mathbf{x}_{1}+c(\mathbf{x}_{2}-\mathbf{x}_{1}))=A\mathbf{x}_{1}+c(A\mathbf{x}_{2}-A\mathbf{x}_{1})=\mathbf{b}+c(\mathbf{b}-\mathbf{b})=\mathbf{b}$$

Therefore, having two solutions automatically creates **infinitely many** solutions, so a singular system can only have **0** or **infinitely many** solutions—never exactly two.
