## 1. State the three elementary row operations and explain why none of them can turn a non-singular matrix into a singular matrix (or vice versa).

### Three Elementary Row Operations

1. **Row Swapping (Interchange):** Interchange two rows of a matrix ($R_i \leftrightarrow R_j$).
2. **Row Scaling:** Multiply all entries of a row by a **nonzero** scalar $c \neq 0$ ($R_i \to cR_i$).
3. **Row Addition (Replacement):** Add a scalar multiple of one row to another row ($R_i \to R_i + kR_j$, where $i \neq j$).

---

### Why None of Them Can Turn a Non-Singular Matrix into a Singular Matrix (or Vice Versa)

A square matrix $A$ is **non-singular** (invertible) if and only if $\det(A) \neq 0$, and **singular** (non-invertible) if and only if $\det(A) = 0$.

#### 1. Determinant Perspective
Each elementary row operation is equivalent to left-multiplying $A$ by an **elementary matrix** $E$, which affects the determinant as follows:

* **Row swap ($R_i \leftrightarrow R_j$):** $\det(EA) = -\det(A)$
* **Row scaling ($R_i \to cR_i$, $c \neq 0$):** $\det(EA) = c \cdot \det(A)$
* **Row addition ($R_i \to R_i + kR_j$):** $\det(EA) = \det(A)$

Because the multiplier ($-1$, $c \neq 0$, or $1$) is **never zero**:
* If $\det(A) \neq 0$, then $\det(EA) \neq 0$ (a non-singular matrix remains **non-singular**).
* If $\det(A) = 0$, then $\det(EA) = 0$ (a singular matrix remains **singular**).

#### 2. Reversibility (Invertibility) Perspective
Every elementary row operation is **reversible** by another elementary row operation:

* Swapping $R_i \leftrightarrow R_j$ is undone by swapping $R_i \leftrightarrow R_j$ again.
* Scaling $R_i \to cR_i$ ($c \neq 0$) is undone by $R_i \to \frac{1}{c}R_i$.
* Replacing $R_i \to R_i + kR_j$ is undone by $R_i \to R_i - kR_j$.

Therefore, every elementary matrix $E$ is invertible ($E^{-1}$ exists). Since the product of two invertible matrices ($EA$) is always invertible—and if $EA$ were invertible then $A = E^{-1}(EA)$ would also be invertible—elementary row operations preserve both non-singularity and singularity.


## 2. Perform the sequence of row operations $R_2 \leftarrow R_2 - 2R_1$ followed by $R_3 \leftarrow R_3 + R_1$ on the matrix:
   $$A = \begin{pmatrix} 1 & -1 & 3 \\ 2 & 3 & 1 \\ -1 & 4 & -2 \end{pmatrix}$$

```python
import sympy as sp

M = sp.Matrix([
    [1,-1,3],
    [2,3,1],
    [-1,4,-2]
])

M = M.elementary_row_op('n->n+km', row=1, k=-2, row2=0)
M = M.elementary_row_op('n->n+km', row=2, k=1, row2=0)

sp.pprint(M)

```

## 3. Let $A$ be a $3 \times 3$ matrix with $\det(A) = 5$. If you swap Row 1 and Row 3, and then multiply Row 2 by $-3$, what is the determinant of the resulting matrix $B$?

**det(B) = 15**

* **Initial:** det(A) = 5
* **Row Swap (R1 <-> R3):** det(A') = (-1) * det(A) = -5
* **Row Scaling (R2 -> -3R2):** det(B) = (-3) * det(A') = (-3)(-5) = 15


## 4. Use elementary row operations to determine if the following matrix is singular or non-singular:
   $$M = \begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{pmatrix}$$

```python
M = sp.Matrix([
    [1,2,3],
    [4,5,6],
    [7,8,9]
])

M = M.elementary_row_op('n->n+km', row1=1, k=-4, row2=0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row1=2, k=sp.Rational(1,7), row2=0 )
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1=2, k=-1, row2=0)
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=1, k=sp.Rational(1,-3))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->kn', row=2, k=sp.Rational(-7,6))
sp.pprint(M)
```

```python
M = M.elementary_row_op('n->n+km', row1=2, k=-1, row2=1)
sp.pprint(M)
```

### Conclusion

Because the third row consists entirely of zeros ($R_3 = \begin{pmatrix} 0 & 0 & 0 \end{pmatrix}$), the matrix has a rank of $2$ (less than its size of $3$) and cannot be reduced to the $3 \times 3$ identity matrix. Therefore, **$M$ is singular**.


## 5. Suppose matrix $B$ is obtained from matrix $A$ by replacing $R_3$ with $R_3 - 4R_2$. Express this operation as multiplication by an elementary matrix $E$ such that $B = EA$, and state $\det(E)$.

To represent the row operation $R_3 \to R_3 - 4R_2$ as an elementary matrix $E$ (assuming a standard $3 \times 3$ matrix), apply the same operation to the $3 \times 3$ identity matrix $I_3$:

$$I_3 = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{bmatrix} \xrightarrow{R_3 \to R_3 - 4R_2} E = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -4 & 1 \end{bmatrix}$$

#### 1. Matrix Equation
Multiplying $A$ on the left by $E$ yields $B$:

$$B = EA = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & -4 & 1 \end{bmatrix} A$$

#### 2. Determinant of $E$
Since $E$ is a lower triangular matrix, its determinant is the product of its main diagonal entries (and adding a multiple of one row to another does not change the determinant):

$$\det(E) = 1 \times 1 \times 1 = 1$$
