## 1. A linear transformation $T: \mathbb{R}^2 \to \mathbb{R}^2$ is defined by the matrix $A = \begin{pmatrix} 3 & 1 \\ 2 & 4 \end{pmatrix}$. If $T$ is applied to the unit square, what is the area of the resulting parallelogram?

```python
import numpy as np

A = np.array([
    [3,1],
    [2,4]
])
```

The unit square has a width of $1$ and a height of $1$, so $\text{Area}_{\text{original}} = 1 \times 1 = 1$.

```python
abs_det_A = np.abs(np.linalg.det(A))
area = abs_det_A * 1
print(np.round(area))
```

## 2. Find the area of the parallelogram in $\mathbb{R}^2$ formed by the column vectors $\mathbf{v}_1 = \begin{pmatrix} -2 \\ 5 \end{pmatrix}$ and $\mathbf{v}_2 = \begin{pmatrix} 3 \\ 1 \end{pmatrix}$.

```python
A = np.array([
    [-2,3],
    [5,1]
])
```

```python
area = np.abs(np.linalg.det(A))
print(area)
```

## 3. A triangle in the coordinate plane has vertices at $(0,0)$, $(4,1)$, and $(2,5)$. Use a $2 \times 2$ determinant to calculate the area of this triangle.

$\text{Area of Triangle} = \frac{1}{2} \times \text{Area of Parallelogram}$

```python
A = np.array([
    [4,1],
    [2,5]
])
```

```python
area = (1/2)*(np.abs(np.linalg.det(A)))
print(np.round(area)) 
```

### 4. A region $S$ in $\mathbb{R}^2$ has an area of $12$ square units. If $S$ is transformed by the matrix $M = \begin{pmatrix} -2 & 0 \\ 3 & 4 \end{pmatrix}$, what is the area of the transformed region, and does the transformation reverse orientation?

```python
A = np.array([
    [-2,0],
    [3,4]
])
```

```python
det_A = np.linalg.det(A)
if det_A < 0:
    print("The area is negative, which means the orientation of the vectors is reversed.")

area = np.abs(det_A) * 12
print(np.round(area))
```
